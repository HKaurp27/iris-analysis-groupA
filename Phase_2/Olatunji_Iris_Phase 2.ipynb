```python
# 1. Variable and Dataset Metada
## Dataset Metadata Configuration

iris_metadata = {
    "dataset_name": "UCI Iris Dataset",
    "n_rows": 150,
    "n_columns": 5,
    "features": ["sepal_length", "sepal_width", "petal_length", "petal_width"],
    "target_classes": ["Setosa", "Versicolor", "Virginica"],
    "units": "cm"
}

# Formatted metadata summary display
print(f"--- {iris_metadata['dataset_name']} Metadata Summary ---")
for key, value in iris_metadata.items():
    print(f". {key.replace('_', ' ').title()}: {value}")
```

```python
#2. Control Structures
##Threshold check with formatted output and Visual status indicators

petal_length = 4.2  # Example value

threshold = 1.5

category = "Large petal" if petal_length > threshold else "Small petal"
indicator = "🟢" if category == "Large petal" else "🔵"

print(
    f"{indicator} Measured Petal Length: {petal_length} cm -> Category: '{category}' (Threshold: {threshold} cm)"
)
```

```python
#2. Control Structures
##Threshold check with formatted output and Visual status indicators
### Prompt the user to enter a petal length measurement

user_input = input("Enter the measured petal length (in cm): ")

# Convert the text input into a decimal number (float)
petal_length = float(user_input)
threshold = 1.5

# Control structure to check length and assign category
category = "Large petal" if petal_length > threshold else "Small petal"

# Visual status indicator mapping
indicator = "🟢" if category == "Large petal" else "🔵"

# Print out the dynamic result
print(
    f"{indicator} Measured Petal Length: {petal_length} cm -> Category: '{category}' (Threshold: {threshold} cm)"
)
```

```python
#3. Loops over Species List

species_list = ["Setosa", "Versicolor", "Virginica"]

print("Botanical Center Tracked Species:")
for index, species in enumerate(species_list, start=1):
    print(
        f"  {index}. Iris {species}"
    )
```

```python
#4. Classification Function (classify_flower)

def classify_flower(sepal_length: float, petal_length: float) -> str:
    """Classifies an Iris flower size category based on sepal and petal measurements.

    Parameters:
        sepal_length (float): Sepal length in centimeters.
        petal_length (float): Petal length in centimeters.

    Returns:
        str: Size classification ('Small', 'Medium', or 'Large').
    """
    # validate input values
    if sepal_length <= 0 or petal_length <= 0:
        return "Invalid Measurement"

    # classification based on petal length
    if petal_length < 2.0:
        return "Small"
    elif 2.0 <= petal_length <= 4.8:
        return "Medium"
    else:
        return "Large"


# Sample values
sample_measurements = [
    (5.1, 1.4),  
    (6.0, 4.2), 
    (6.9, 5.4), 
]

print("--- Function Demonstration ---")
for sepal, petal in sample_measurements:
    result = classify_flower(sepal, petal)
    print(
        f"Sepal / Petal Length Input {sepal} cm, {petal} cm  =>  Classification: {result}"
    )
```
