```python
#Functions
# We define a function using 'def', giving it a name and inputs in parentheses: sepal_length and petal_length
def classify_flower(sepal_length, petal_length):
    # Check if measurements fall into the small category criteria
    if petal_length < 2.0 and sepal_length < 5.0:
        return "small"
    
    # Check if measurements fall into the large category criteria
    elif petal_length >= 5.0 and sepal_length >= 6.0:
        return "large"
        
    # If neither of the above matches, it defaults to medium
    else:
        return "medium"

# Let's test (call) our function with different measurements!
flower_type_1 = classify_flower(4.8, 1.4)
flower_type_2 = classify_flower(6.7, 5.5)

print("Flower 1 category:", flower_type_1)  
print("Flower 2 category:", flower_type_2)  
```

# Phase 3-File Handling & Data Cleaning

```python
import pandas as pd

# Define the columns for the raw UCI dataset
column_names = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'species']

# Load the raw file (make sure 'iris.data' is the actual name of your extracted file)
df = pd.read_csv('iris.data', header=None, names=column_names)

# Save it as iris.csv in the same folder
df.to_csv('iris.csv', index=False)

print("iris.csv has been successfully created!")
```

```python
import csv

# Define our input file name and the output summary text file name
input_file = 'iris.csv'
output_summary = 'summary.txt'

# Create counter variables to store our data as we read through the file
row_count = 0
sepal_lengths = []  # An empty list to store all sepal length numbers

# Open the CSV file in read mode ('r')
with open(input_file, mode='r') as file:
    csv_reader = csv.reader(file)
    
    # Skip the header row (column names) so we only count actual flower data
    header = next(csv_reader)
    
    # Loop through every single row in the CSV file one by one
    for row in csv_reader:
        if row:  # Make sure the row is not empty
            row_count += 1  # Add 1 to our running row counter
            
            # Grab the first column (index 0 is sepal_length), convert it to a decimal number (float), and add it to our list
            sepal_lengths.append(float(row[0]))

# Calculate statistics using Python's built-in math functions
min_sepal = min(sepal_lengths)                    # Find the smallest sepal length
max_sepal = max(sepal_lengths)                    # Find the largest sepal length
avg_sepal = sum(sepal_lengths) / len(sepal_lengths) # Calculate the average (mean)

# Open a new text file in write mode ('w') to save our summary results
with open(output_summary, mode='w') as summary_file:
    summary_file.write("--- Iris Dataset Summary ---\n")
    summary_file.write(f"Total Rows (excluding header): {row_count}\n")
    summary_file.write(f"Minimum Sepal Length: {min_sepal}\n")
    summary_file.write(f"Maximum Sepal Length: {max_sepal}\n")
    summary_file.write(f"Average Sepal Length: {avg_sepal:.2f}\n")

print("Success! 'summary.txt' has been created.")
```

# Phase 4-Analysis & Visualization

```python
# ---This code loops through your dataset row-by-row to count how many flowers belong to each species.

# 1. Count Flowers Manually Using a for Loop 
import pandas as pd

# Load the cleaned dataset
df = pd.read_csv('iris_cleaned.csv')

# Create an empty dictionary to keep track of counts
species_counts = {}

# Loop through each species entry in the column
for species in df['species']:
    if species in species_counts:
        species_counts[species] += 1
    else:
        species_counts[species] = 1
        
# Print the final counts
print("Flower Count per Species")
for species, count in species_counts.items():
    print(f"{species}: {count} flowers")
```

```python
# 2. Categorize Flowers Using If/Else & Save to CSV
#---This code checks the petal length of every flower using if-elif-else control structures, adds a category label, and saves the result to a new file named iris_classes.csv

# Create an empty list to store the category labels
categories = []

# Loop through every petal length in the dataset
for petal_len in df['petal_length']:
    if petal_len < 2:
        categories.append("Short Petal")
    elif 2 <= petal_len <= 5:
        categories.append("Medium Petal")
    else:
        categories.append("Long Petal")

# Add the new list as a column in our DataFrame
df['petal_category'] = categories

# Save the updated data into 'iris_classes.csv'
df.to_csv('iris_classes.csv', index=False)

print("Success! Categorized data saved to 'iris_classes.csv'.")
print(df[['petal_length', 'petal_category']].head(10))
```

```python
# 3. Pandas Statistical Analysis
#---This code calculates the mean, median, minimum, and maximum values, and groups the data by species to compute their averages.

# Calculate mean, median, min, and max for the numerical columns
stats_summary = df[['sepal_length', 'sepal_width', 'petal_length', 'petal_width']].agg(['mean', 'median', 'min', 'max'])

print("Overall Statistics")
print(stats_summary)

# Group by species and compute average measurements
species_averages = df.groupby('species').mean(numeric_only=True)

print("\nAverage Measurements by Species")
print(species_averages)
```

```python
# 4.Data Visualizations (Plots & Charts)

# 1. Histogram of Petal Lengths
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

plt.figure(figsize=(8, 5))
sns.histplot(df['petal_length'], bins=20, kde=True, color='purple')
plt.title('Histogram of Petal Lengths')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Count')
plt.show()
```

```python
# 2. Bar Chart of Average Sepal Length per Species
plt.figure(figsize=(8, 5))
species_averages['sepal_length'].plot(kind='bar', color=['skyblue', 'salmon', 'lightgreen'])
plt.title('Average Sepal Length per Species')
plt.xlabel('Species')
plt.ylabel('Average Sepal Length (cm)')
plt.xticks(rotation=0)
plt.show()
```

```python
# 3. Boxplot of Petal Lengths by Species
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

# FIXED: Added hue='species', legend=False, and corrected column name to 'petal_length'
sns.boxplot(
    data=df, 
    x='species', 
    y='petal_length', 
    hue='species', 
    palette='Set2', 
    legend=False
)

plt.title('Boxplot of Petal Lengths by Species')
plt.xlabel('Species')
plt.ylabel('Petal Length (cm)')
plt.show()
```

```python
# 4. Scatter Plot (Sepal Length vs Sepal Width)
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='sepal_length', y='sepal_width', hue='species', palette='Set1', s=100)
plt.title('Sepal Length vs Sepal Width')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')
plt.legend(title='Species')
plt.show()
```
