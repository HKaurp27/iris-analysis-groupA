```python
# 4. Function
# Classify flower size using sepal and petal length

def classify_flower(sepal_length, petal_length):
    if sepal_length <= 0 or petal_length <= 0:
        return "Invalid Measurement"

    if sepal_length < 5.0 and petal_length < 2.0:
        return "Small"
    elif sepal_length <= 6.5 and petal_length <= 4.8:
        return "Medium"
    else:
        return "Large"


sample_measurements = [
    (4.9, 1.4),
    (6.0, 4.2),
    (6.9, 5.4)
]

for sepal, petal in sample_measurements:
    result = classify_flower(sepal, petal)
    print("Sepal:", sepal, "Petal:", petal, "Classification:", result)
```

```python
import csv

with open("Downloads/iris/iris.data", "r") as file:
    reader = csv.reader(file)
    iris_data = list(reader)

iris_data = [row for row in iris_data if row]

print("Number of rows:", len(iris_data))

for row in iris_data[:5]:
    print(row)
```

```python
# Get the number of rows
row_count = len(iris_data)

# Make lists for each measurement
sepal_length = []
sepal_width = []
petal_length = []
petal_width = []

# Add the numbers from the dataset into the lists
for row in iris_data:
    sepal_length.append(float(row[0]))
    sepal_width.append(float(row[1]))
    petal_length.append(float(row[2]))
    petal_width.append(float(row[3]))

# Print row count
print("Row count:", row_count)

# Print minimum, maximum, and average
print("Sepal Length")
print("Minimum:", min(sepal_length))
print("Maximum:", max(sepal_length))
print("Average:", sum(sepal_length) / row_count)

print()

print("Sepal Width")
print("Minimum:", min(sepal_width))
print("Maximum:", max(sepal_width))
print("Average:", sum(sepal_width) / row_count)

print()

print("Petal Length")
print("Minimum:", min(petal_length))
print("Maximum:", max(petal_length))
print("Average:", sum(petal_length) / row_count)

print()

print("Petal Width")
print("Minimum:", min(petal_width))
print("Maximum:", max(petal_width))
print("Average:", sum(petal_width) / row_count)
```

```python
# Create a new text file called summary.txt
with open("summary.txt", "w") as file:

    # Write the title
    file.write("Iris Dataset Summary\n")
    file.write("\n")

    # Write the total number of rows
    file.write("Row Count: " + str(row_count) + "\n")
    file.write("\n")

    # Sepal length summary
    file.write("Sepal Length\n")
    file.write("Minimum: " + str(min(sepal_length)) + "\n")
    file.write("Maximum: " + str(max(sepal_length)) + "\n")
    file.write("Average: " + str(round(sum(sepal_length) / row_count, 2)) + "\n")
    file.write("\n")

    # Sepal width summary
    file.write("Sepal Width\n")
    file.write("Minimum: " + str(min(sepal_width)) + "\n")
    file.write("Maximum: " + str(max(sepal_width)) + "\n")
    file.write("Average: " + str(round(sum(sepal_width) / row_count, 2)) + "\n")
    file.write("\n")

    # Petal length summary
    file.write("Petal Length\n")
    file.write("Minimum: " + str(min(petal_length)) + "\n")
    file.write("Maximum: " + str(max(petal_length)) + "\n")
    file.write("Average: " + str(round(sum(petal_length) / row_count, 2)) + "\n")
    file.write("\n")

    # Petal width summary
    file.write("Petal Width\n")
    file.write("Minimum: " + str(min(petal_width)) + "\n")
    file.write("Maximum: " + str(max(petal_width)) + "\n")
    file.write("Average: " + str(round(sum(petal_width) / row_count, 2)) + "\n")

# Let us know the file was created
print("summary.txt created successfully")
```
