```python
# 4. Function
# Classify flower size using sepal and petal length

def classify_flower(sepal_length, petal_length):
    if sepal_length <= 0 or petal_length <= 0:
        return "Invalid Measurement"

    if sepal_length < 5.0 and petal_length < 2.0:
        return "Small"
    elif sepal_length <= 6.5 and petal_length <= 4.8:
        return "Medium"
    else:
        return "Large"


sample_measurements = [
    (4.9, 1.4),
    (6.0, 4.2),
    (6.9, 5.4)
]

for sepal, petal in sample_measurements:
    result = classify_flower(sepal, petal)
    print("Sepal:", sepal, "Petal:", petal, "Classification:", result)
```

Phase 3 File Handling & Data Cleaning

```python
import csv

with open("Downloads/iris/iris.data", "r") as file:
    reader = csv.reader(file)
    iris_data = list(reader)

iris_data = [row for row in iris_data if row]

print("Number of rows:", len(iris_data))

for row in iris_data[:5]:
    print(row)
```

```python
# Get the number of rows
row_count = len(iris_data)

# Make lists for each measurement
sepal_length = []
sepal_width = []
petal_length = []
petal_width = []

# Add the numbers from the dataset into the lists
for row in iris_data:
    sepal_length.append(float(row[0]))
    sepal_width.append(float(row[1]))
    petal_length.append(float(row[2]))
    petal_width.append(float(row[3]))

# Print row count
print("Row count:", row_count)

# Print minimum, maximum, and average
print("Sepal Length")
print("Minimum:", min(sepal_length))
print("Maximum:", max(sepal_length))
print("Average:", sum(sepal_length) / row_count)

print()

print("Sepal Width")
print("Minimum:", min(sepal_width))
print("Maximum:", max(sepal_width))
print("Average:", sum(sepal_width) / row_count)

print()

print("Petal Length")
print("Minimum:", min(petal_length))
print("Maximum:", max(petal_length))
print("Average:", sum(petal_length) / row_count)

print()

print("Petal Width")
print("Minimum:", min(petal_width))
print("Maximum:", max(petal_width))
print("Average:", sum(petal_width) / row_count)
```

```python
# Create a new text file called summary.txt
with open("summary.txt", "w") as file:

    # Write the title
    file.write("Iris Dataset Summary\n")
    file.write("\n")

    # Write the total number of rows
    file.write("Row Count: " + str(row_count) + "\n")
    file.write("\n")

    # Sepal length summary
    file.write("Sepal Length\n")
    file.write("Minimum: " + str(min(sepal_length)) + "\n")
    file.write("Maximum: " + str(max(sepal_length)) + "\n")
    file.write("Average: " + str(round(sum(sepal_length) / row_count, 2)) + "\n")
    file.write("\n")

    # Sepal width summary
    file.write("Sepal Width\n")
    file.write("Minimum: " + str(min(sepal_width)) + "\n")
    file.write("Maximum: " + str(max(sepal_width)) + "\n")
    file.write("Average: " + str(round(sum(sepal_width) / row_count, 2)) + "\n")
    file.write("\n")

    # Petal length summary
    file.write("Petal Length\n")
    file.write("Minimum: " + str(min(petal_length)) + "\n")
    file.write("Maximum: " + str(max(petal_length)) + "\n")
    file.write("Average: " + str(round(sum(petal_length) / row_count, 2)) + "\n")
    file.write("\n")

    # Petal width summary
    file.write("Petal Width\n")
    file.write("Minimum: " + str(min(petal_width)) + "\n")
    file.write("Maximum: " + str(max(petal_width)) + "\n")
    file.write("Average: " + str(round(sum(petal_width) / row_count, 2)) + "\n")

# Let us know the file was created
print("summary.txt created successfully")
```

# Phase 4 – Analysis & Visualization

```python

# Count how many flowers of each species using a for loop

setosa_count = 0
versicolor_count = 0
virginica_count = 0

for species in df["species"]:
    if species == "Iris-setosa":
        setosa_count += 1
    elif species == "Iris-versicolor":
        versicolor_count += 1
    elif species == "Iris-virginica":
        virginica_count += 1

print("Iris-setosa:", setosa_count)
print("Iris-versicolor:", versicolor_count)
print("Iris-virginica:", virginica_count)
```

```python
# Categorize flowers based on petal length

def categorize_petal(petal_length):
    if petal_length < 2:
        return "Short Petal"
    elif petal_length <= 5:
        return "Medium Petal"
    else:
        return "Long Petal"

df["petal_category"] = df["petal_length"].apply(categorize_petal)

df[["petal_length", "petal_category"]].head(10)
```

```python
# Calculate mean, median, minimum, and maximum

print("Sepal Length")
print("Mean:", df["sepal_length"].mean())
print("Median:", df["sepal_length"].median())
print("Minimum:", df["sepal_length"].min())
print("Maximum:", df["sepal_length"].max())

print()

print("Petal Length")
print("Mean:", df["petal_length"].mean())
print("Median:", df["petal_length"].median())
print("Minimum:", df["petal_length"].min())
print("Maximum:", df["petal_length"].max())
```

# 4. Group by species and calculate averages

```python
# Group by species and calculate averages

species_averages = df.groupby("species")[
    ["sepal_length", "sepal_width", "petal_length", "petal_width"]
].mean()

print(species_averages)
```

# 5. Histogram of petal lengths

```python
import matplotlib.pyplot as plt

# Histogram of petal lengths
plt.hist(df["petal_length"], bins=20, edgecolor="black")

plt.xlabel("Petal Length")
plt.ylabel("Frequency")
plt.title("Histogram of Petal Lengths")

plt.show()
```

```python
# Bar chart of average sepal length per species

average_sepal = df.groupby("species")["sepal_length"].mean()


average_sepal.plot(
    kind="bar",
    color=["skyblue", "orange", "gray"]
)

plt.xlabel("Species")
plt.ylabel("Average Sepal Length")
plt.title("Average Sepal Length by Species")
plt.xticks(rotation=0)

plt.show()
```

```python
# Boxplot of petal length by species

df.boxplot(column="petal_length", by="species")

plt.xlabel("Species")
plt.ylabel("Petal Length")
plt.title("Petal Length by Species")
plt.suptitle("")

plt.show()
```

Discussion questions:

1. Which species has the smallest range of petal length?
Iris-setosa has the smallest range of petal lengths.

2. Which species shows the largest spread in petal length values?

Iris-virginica has the largest spread in petal length values.

3.How do the medians of the three species compare? What does this suggest about their flower size? 

The median petal length is lowest for Iris-setosa, higher for Iris-versicolor, and highest for Iris-virginica. This suggests that setosa generally has the smallest petals, versicolor has medium-sized petals, and virginica has the largest petals.

```python

# Scatter plot by species

for species in df["species"].unique():
    group = df[df["species"] == species]

    plt.scatter(
        group["sepal_length"],
        group["sepal_width"],
        label=species
    )

plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')
plt.title('Sepal Length and Width Correlation ')
plt.legend()

plt.show()
```

Discussion questions:

1. Do sepal length and sepal width appear strongly correlated? Explain why or why not using evidence from the scatter plot.

Sepal length and sepal width do not appear to be strongly correlated. The points are spread out rather than forming a clear straight-line pattern, so there is not a strong relationship between the two measurements.

2. Based on the scatter plot, boxplot, and histogram, which flower features(sepal length, sepal width, petal length, petal width) seem most useful for separating the species?

Petal length and petal width seem to be the most useful features for separating the Iris species. Iris-setosa has much smaller petal measurements, Iris-versicolor has medium-sized petals, and Iris-virginica has the largest petals. Sepal length helps somewhat, but sepal width overlaps more between the species.
