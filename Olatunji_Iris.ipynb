```python
#4. Classification Function (classify_flower)

def classify_flower(sepal_length: float, petal_length: float) -> str:
    """Classifies an Iris flower size category based on sepal and petal measurements.

    Parameters:
        sepal_length (float): Sepal length in centimeters.
        petal_length (float): Petal length in centimeters.

    Returns:
        str: Size classification ('Small', 'Medium', or 'Large').
    """
    # validate input values
    if sepal_length <= 0 or petal_length <= 0:
        return "Invalid Measurement"

    # classification based on petal length
    if petal_length < 2.0:
        return "Small"
    elif 2.0 <= petal_length <= 4.8:
        return "Medium"
    else:
        return "Large"


# Sample values
sample_measurements = [
    (5.1, 1.4),  
    (6.0, 4.2), 
    (6.9, 5.4), 
]

print("--- Function Demonstration ---")
for sepal, petal in sample_measurements:
    result = classify_flower(sepal, petal)
    print(
        f"Sepal / Petal Length Input {sepal} cm, {petal} cm  =>  Classification: {result}"
    )
```

##Manual Species Counting with a Loop

```python
# I initialize a dictionary to keep track of species counts
species_counts = {"Iris-setosa": 0, "Iris-versicolor": 0, "Iris-virginica": 0}

# Loop through the species column manually
for species in df["class"]:
    if species in species_counts:
        species_counts[species] += 1

print("--- Manual Species Count using a Loop ---")
for species, count in species_counts.items():
    print(f"• {species}: {count} flowers")
```

##Custom Flower Categorization (If/Else)

```python
def Petal_Categorize(petal_length):
    if petal_length < 2:
        return "Short Petal"
    elif 2 <= petal_length <= 5:
        return "Medium Petal"
    else:
        return "Long Petal"

# Apply the categorization function to create a new column in your DataFrame
df["Petal_Categorize"] = df["petal_length"].apply(Petal_Categorize)

print("--- Flower Categories Preview ---")
print(df[["petal_length", "Petal_Categorize"]].head(150))
#df.to_csv('iris_classes.csv', index = False, header=True)
```

```python
# Compute correlation matrix to back up your findings in the report
correlation_matrix = df[["sepal_length", "sepal_width", "petal_length", "petal_width"]].corr()

print("--- Feature Correlation Matrix ---")
print(correlation_matrix)
```

```python

# To Set up the correlation graph size
plt.figure(figsize=(8, 6))

# Generate the heatmap
sns.heatmap(
    correlation_matrix, 
    annot=True,         # Show the actual correlation values inside the cells
    cmap="coolwarm",    # Color map (coolwarm works great for correlations: blue=negative, red=positive)
    fmt=".2f",          # Format numbers to 2 decimal places
    linewidths=0.5,     # Add grid lines between cells
    cbar=True           # Show the color intensity scale bar on the side
)

plt.title("Iris Feature Correlation Heatmap", fontsize=14, pad=15)
plt.savefig("iris_correlation_heatmap.png", bbox_inches="tight")
plt.show()
```
